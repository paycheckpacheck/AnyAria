# LLC_RESONANT_TANK Block Simulation

This notebook derives and verifies the LLC resonant tank component values.

## Design Requirements
- Resonant frequency: fr = 250kHz
- Quality factor: Q = 0.3-0.5
- Input voltage: 48V (half-bridge output 0-48V)
- Output power: 120W
- Resonant current: ~8A RMS at full load

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Constants
fr_target = 250e3  # Target resonant frequency (Hz)
Vin = 48  # Input voltage (V)
Pout = 120  # Output power (W)
Vout = 12  # Output voltage (V)
Iout = Pout / Vout  # Output current (A)
n_turns = 4  # Transformer turns ratio (primary:secondary)

print(f"Design targets:")
print(f"  fr = {fr_target/1e3:.0f} kHz")
print(f"  Vin = {Vin} V")
print(f"  Pout = {Pout} W")
print(f"  Iout = {Iout} A")
```

## Resonant Tank Component Calculations

### Resonant Frequency
$$f_r = \frac{1}{2\pi\sqrt{L_r \times C_r}}$$

### Quality Factor
$$Q = \frac{\sqrt{L_r/C_r}}{R_{ac}}$$

where $R_{ac}$ is the equivalent AC load resistance reflected to primary.

```python
# Initial values (will be optimized)
Lr_initial = 25e-6  # Resonant inductance (H)
Cr_initial = 100e-9  # Resonant capacitance (F)

# Calculate initial resonant frequency
fr_initial = 1 / (2 * np.pi * np.sqrt(Lr_initial * Cr_initial))

print(f"Initial component values:")
print(f"  Lr = {Lr_initial*1e6:.1f} µH")
print(f"  Cr = {Cr_initial*1e9:.0f} nF")
print(f"  fr(calculated) = {fr_initial/1e3:.0f} kHz")
print(f"  Target fr = {fr_target/1e3:.0f} kHz")
print(f"  Error: {(fr_initial-fr_target)/fr_target*100:.1f}%")
```
