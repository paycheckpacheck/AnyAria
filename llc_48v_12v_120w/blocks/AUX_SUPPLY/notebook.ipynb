## Design Calculations

### TPS5430 Output Voltage

$$V_{OUT} = V_{REF} \times \left(1 + \frac{R_{FB1}}{R_{FB2}}\right) = 0.8V \times \left(1 + \frac{10k\Omega}{2.43k\Omega}\right) = 4.11V$$

**Action needed:** Adjust R3 to achieve exactly 5.0V

$$R_{FB2} = \frac{R_{FB1} \times V_{REF}}{V_{OUT} - V_{REF}} = \frac{10k\Omega \times 0.8V}{5.0V - 0.8V} = 1.90k\Omega$$

**Standard value:** 1.91kΩ (1% E96 series) → VOUT = 5.01V

## What Is Not Verified

1. **Buck converter loop stability** - Phase/gain margins with actual compensation network
2. **Startup behavior** - Inrush current, soft-start timing
3. **Thermal analysis** - Junction temperatures under full load
4. **Load transient response** - Voltage dip/overshoot on 0-500mA load steps
5. **EMI/switching noise** - 500kHz fundamental and harmonics
6. **Cross-regulation** - Effect of varying loads on other rails
7. **RP2040 operation at 1.2V DVDD** - Whether centered voltage (vs 1.1V nominal) affects performance
